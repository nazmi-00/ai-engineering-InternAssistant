```python
import re
from pathlib import Path

INPUT_PATH = Path(
    "/content/week3_project/data/processed/GarisPanduanLI_extracted.txt"
)

OUTPUT_PATH = Path(
    "/content/week3_project/data/processed/GarisPanduanLI_cleaned.txt"
)

with open(INPUT_PATH, "r", encoding="utf-8") as file:
    raw_text = file.read()

# Remove unwanted null characters
cleaned_text = raw_text.replace("\x00", "")

# Normalize spaces and tabs
cleaned_text = re.sub(r"[ \t]+", " ", cleaned_text)

# Remove excessive blank lines
cleaned_text = re.sub(r"\n{3,}", "\n\n", cleaned_text)

# Preserve page markers and paragraph structure
cleaned_text = cleaned_text.strip()

with open(OUTPUT_PATH, "w", encoding="utf-8") as file:
    file.write(cleaned_text)

print("Cleaning completed!")
print("Original characters:", len(raw_text))
print("Cleaned characters:", len(cleaned_text))
print("Output file:", OUTPUT_PATH)
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
from pathlib import Path

INPUT_PATH = Path(
    "/content/week3_project/data/processed/GarisPanduanLI_cleaned.txt"
)

OUTPUT_PATH = Path(
    "/content/week3_project/data/processed/GarisPanduanLI_chunks_v2.txt"
)

with open(INPUT_PATH, "r", encoding="utf-8") as file:
    text = file.read()

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1200,
    chunk_overlap=100,
    separators=["\n\n", "\n", ". ", " ", ""]
)

chunks_v2 = text_splitter.split_text(text)

print("Total chunks:", len(chunks_v2))

with open(OUTPUT_PATH, "w", encoding="utf-8") as file:
    for i, chunk in enumerate(chunks_v2):
        file.write(f"\n--- CHUNK {i+1} ---\n")
        file.write(chunk)
        file.write("\n")

print("Saved:", OUTPUT_PATH)
```

```python
import json
import re
import hashlib
from pathlib import Path
from datetime import datetime, timezone

# Define file paths
BASE_DIR = Path("/content/week3_project")
RAW_DIR = BASE_DIR / "data" / "raw"
PROCESSED_DIR = BASE_DIR / "data" / "processed"

PDF_PATH = RAW_DIR / "GarisPanduanLI.pdf"
OUTPUT_PATH = PROCESSED_DIR / "GarisPanduanLI_chunks_metadata.json"

# Use the selected Version 2 chunks
metadata_records = []
last_page = 1

for i, chunk in enumerate(chunks_v2):

    # Identify page numbers from page markers
    page_markers = re.findall(r"--- PAGE (\d+) ---", chunk)

    if page_markers:
        pages = sorted(set(int(p) for p in page_markers))
        last_page = pages[-1]
    else:
        pages = [last_page]

    # Identify section headings
    section_matches = re.findall(
        r"(?m)^\s*(\d+(?:\.\d+)+)\s*\n\s*([^\n]+)",
        chunk
    )

    sections = [
        f"{number} {heading.strip()}"
        for number, heading in section_matches
    ]

    # Create metadata record
    record = {
        "chunk_id": f"GLI_{i+1:03d}",
        "source": PDF_PATH.name,
        "pages": pages,
        "sections": sections,
        "language": "Malay",
        "chunking_method": "RecursiveCharacterTextSplitter",
        "chunk_size": 1200,
        "chunk_overlap": 100,
        "actual_characters": len(chunk),
        "text": chunk.strip()
    }

    metadata_records.append(record)

# Save metadata as JSON
with open(OUTPUT_PATH, "w", encoding="utf-8") as file:
    json.dump(metadata_records, file, ensure_ascii=False, indent=4)

print("Metadata creation completed!")
print("Total metadata records:", len(metadata_records))
print("Saved to:", OUTPUT_PATH)
```

```python
import hashlib
import json
from datetime import datetime, timezone

def get_file_hash(file_path):
    sha256 = hashlib.sha256()

    with open(file_path, "rb") as file:
        for block in iter(lambda: file.read(4096), b""):
            sha256.update(block)

    return sha256.hexdigest()

lineage = {
    "source_file": "GarisPanduanLI.pdf",
    "source_path": str(PDF_PATH),
    "source_sha256": get_file_hash(PDF_PATH),
    "processing_steps": [
        "PDF text extraction using PyMuPDF",
        "Text cleaning",
        "Recursive character text splitting",
        "Metadata generation"
    ],
    "chunking_configuration": {
        "method": "RecursiveCharacterTextSplitter",
        "chunk_size": 1200,
        "chunk_overlap": 100
    },
    "total_chunks": len(chunks_v2),
    "output_file": OUTPUT_PATH.name,
    "generated_at": datetime.now(timezone.utc).isoformat()
}

LINEAGE_PATH = PROCESSED_DIR / "GarisPanduanLI_lineage.json"

with open(LINEAGE_PATH, "w", encoding="utf-8") as file:
    json.dump(lineage, file, ensure_ascii=False, indent=4)

print("Data lineage saved!")
print(LINEAGE_PATH)
```
